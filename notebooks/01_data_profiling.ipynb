Conversión de columnas temporales a tipo datetime
Las columnas relacionadas con fechas y horas se encontraban originalmente almacenadas como tipo str. Para poder realizar operaciones temporales, comparaciones cronológicas y cálculos de duración, se convirtieron al tipo datetime utilizando pd.to_datetime().

Se transformaron las siguientes columnas: order_purchase_timestamp, order_approved_at, order_delivered_carrier_date, order_delivered_customer_date y order_estimated_delivery_date.

Se utilizó el parámetro errors="coerce" para que cualquier valor no interpretable como fecha sea convertido en NaT, preservando así los valores faltantes dentro de un tipo de dato temporal adecuado. Luego de la conversión, se verificó que las cinco columnas quedaran almacenadas como datetime64[us].

Esta transformación permite continuar con controles de calidad más avanzados, como verificar el orden lógico de los eventos del pedido y calcular tiempos transcurridos entre las distintas etapas del proceso.

```python
orders["order_status"].value_counts()

# verificacion de la coherencia cronologica de las fechas, por ejemplo, que la fecha de aprobacion no sea posterior a la fecha de entrega al cliente
# purchase <= approved <= carrier <= customer
#regla 1 purchase <= approved


approved_before_purchase = orders[
    (orders["order_approved_at"] < orders["order_purchase_timestamp"])
]
approved_before_purchase.shape[0]
#el resultado es 0, lo que significa que no hay registros donde la fecha de aprobación sea anterior a la fecha de compra, lo cual es coherente.

#regla 2 approved <= carrier
carrier_before_approved = orders[
    (orders["order_delivered_carrier_date"] < orders["order_approved_at"])
]
carrier_before_approved.shape[0]
#el resultado es 1359, lo que significa que hay 1359 registros donde la fecha de entrega por el transportista sea anterior a la fecha de aprobación, lo cual es incoherente.

#regla 3 carrier <= customer
customer_before_carrier = orders[
    (orders["order_delivered_customer_date"] < orders["order_delivered_carrier_date"])
]
customer_before_carrier.shape[0]
#el resultado es 23, lo que significa que hay 23 registros donde la fecha de entrega al cliente sea anterior a la fecha de entrega por el transportista, lo cual es incoherente.


carrier_before_approved = orders[
    orders["order_delivered_carrier_date"] < orders["order_approved_at"]
].copy()

carrier_before_approved["time_difference"] = (
    carrier_before_approved["order_approved_at"]
    - carrier_before_approved["order_delivered_carrier_date"]
)
carrier_before_approved["time_difference"].describe()
```


Se verificó la coherencia cronológica entre las principales fechas del proceso de los pedidos, considerando como secuencia esperada purchase <= approved <= carrier <= customer. No se encontraron pedidos cuya fecha de aprobación fuera anterior a la fecha de compra. Sin embargo, se detectaron 1359 registros donde la fecha de entrega al transportista es anterior a la aprobación del pedido y 23 registros donde la fecha de entrega al cliente es anterior a la fecha registrada de entrega al transportista. Estos casos se consideran anomalías cronológicas potenciales y requieren un análisis adicional de las diferencias temporales antes de clasificarlos definitivamente como errores de calidad de datos.

Se identificaron 1359 pedidos cuya fecha de entrega al transportista es anterior a la fecha de aprobación. Las diferencias presentan una alta dispersión: van desde 21 segundos hasta aproximadamente 171 días. La mediana es de aproximadamente 17 horas y el 75 % de los casos presenta una diferencia inferior a aproximadamente 1 día y 2 horas. La existencia de valores extremos considerablemente mayores sugiere que las anomalías podrían tener distintas causas, por lo que es necesario analizar la distribución de las diferencias antes de determinar su tratamiento.

```python
carrier_before_approved["difference_hours"] = (
    carrier_before_approved["time_difference"].dt.total_seconds() / 3600
)

print(
    "Menos de 1 hora:",
    (carrier_before_approved["difference_hours"] < 1).sum()
)

print(
    "Entre 1 y 24 horas:",
    (
        (carrier_before_approved["difference_hours"] >= 1) &
        (carrier_before_approved["difference_hours"] < 24)
    ).sum()
)

print(
    "Entre 1 y 7 días:",
    (
        (carrier_before_approved["difference_hours"] >= 24) &
        (carrier_before_approved["difference_hours"] < 168)
    ).sum()
)

print(
    "7 días o más:",
    (carrier_before_approved["difference_hours"] >= 168).sum()
)

carrier_before_approved[
    [
        "order_id",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "time_difference"
    ]
].sort_values(
    "time_difference",
    ascending=False
).head(10)

customer_before_carrier = orders[
    orders["order_delivered_customer_date"] <
    orders["order_delivered_carrier_date"]
].copy()

customer_before_carrier["time_difference"] = (
    customer_before_carrier["order_delivered_carrier_date"]
    - customer_before_carrier["order_delivered_customer_date"]
)

customer_before_carrier["time_difference"].describe()
```

Se encontraron 1359 registros donde order_delivered_carrier_date es anterior a order_approved_at. Al analizar la magnitud de las diferencias:

```python
orders_analysis = orders_clean.merge(
    customers_clean[
        [
            "customer_id",
            "customer_unique_id",
            "customer_city",
            "customer_state"
        ]
    ],
    on="customer_id",
    how="left"
)

print("Filas:", orders_analysis.shape[0])
print("Order ID únicos:", orders_analysis["order_id"].nunique())

orders_analysis["purchase_year"] = (
    orders_analysis["order_purchase_timestamp"].dt.year
)

orders_analysis["purchase_month"] = (
    orders_analysis["order_purchase_timestamp"].dt.month
)

orders_analysis["purchase_year_month"] = (
    orders_analysis["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

orders_analysis["delivery_time_days"] = (
    orders_analysis["order_delivered_customer_date"]
    - orders_analysis["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders_analysis["delivery_delay_days"] = (
    orders_analysis["order_delivered_customer_date"]
    - orders_analysis["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

orders_analysis["delivered_late"] = (
    orders_analysis["delivery_delay_days"] > 0
)

orders_analysis["delivered_late"] = pd.NA

mask_delivered_date = (
    orders_analysis["order_delivered_customer_date"].notna()
)

orders_analysis.loc[
    mask_delivered_date,
    "delivered_late"
] = (
    orders_analysis.loc[
        mask_delivered_date,
        "delivery_delay_days"
    ] > 0
)

items_by_order = (
    order_items_clean
    .groupby("order_id")
    .agg(
        items_count=("order_item_id", "count"),
        products_count=("product_id", "nunique"),
        sellers_count=("seller_id", "nunique"),
        products_value=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
    .reset_index()
)

orders_analysis = orders_analysis.merge(
    items_by_order,
    on="order_id",
    how="left"
)

print("Filas:", orders_analysis.shape[0])
print("Order ID únicos:", orders_analysis["order_id"].nunique())

payments_by_order = (
    payments_clean
    .groupby("order_id")
    .agg(
        payment_value=("payment_value", "sum"),
        payment_records=("payment_sequential", "count"),
        max_installments=("payment_installments", "max"),
        payment_types_count=("payment_type", "nunique")
    )
    .reset_index()
)

orders_analysis = orders_analysis.merge(
    payments_by_order,
    on="order_id",
    how="left"
)

print(orders_analysis.shape[0])
print(orders_analysis["order_id"].nunique())


orders_analysis["has_items"] = (
    orders_analysis["items_count"].notna()
)

orders_analysis["has_payment"] = (
    orders_analysis["payment_value"].notna()
)



reviews_by_order = (
    reviews_clean
    .groupby("order_id")
    .agg(
        review_score_mean=("review_score", "mean"),
        review_count=("review_id", "count"),
        review_comments_count=(
            "review_comment_message",
            lambda x: x.notna().sum()
        )
    )
    .reset_index()
)

orders_analysis = orders_analysis.merge(
    reviews_by_order,
    on="order_id",
    how="left"
)

orders_analysis["has_review"] = (
    orders_analysis["review_count"].notna()
)

print("Filas:", orders_analysis.shape[0])
print(
    "Order ID únicos:",
    orders_analysis["order_id"].nunique()
)

orders_analysis.info()

orders_analysis[
    [
        "order_id",
        "customer_unique_id",
        "order_status",
        "items_count",
        "products_value",
        "freight_value",
        "payment_value",
        "delivery_time_days",
        "delivery_delay_days",
        "delivered_late"
    ]
].head(10)
```

# Construcción de tablas analíticas

Luego de finalizar la auditoría y preparación de los datasets originales de Olist, se construyeron tablas analíticas con distintas granularidades para facilitar el análisis exploratorio y evitar problemas de duplicación de registros durante los `merge`.

La principal decisión de modelado fue no combinar directamente todas las tablas en un único DataFrame, ya que datasets como `order_items`, `payments` y `reviews` poseen múltiples registros asociados a un mismo pedido. Un `merge` directo entre estas tablas podría multiplicar filas y producir sobreconteos en métricas como ventas, pagos o cantidad de pedidos.

Por este motivo, se construyeron tres tablas analíticas principales.

## `orders_analysis`

Granularidad:

`1 fila = 1 pedido`

La tabla parte de `orders` y se complementa con información agregada de clientes, ítems, pagos y reviews.

Incluye, entre otras variables:

- Identificación del pedido y del cliente.
- Estado del pedido.
- Ciudad y estado del cliente.
- Fechas principales del proceso.
- Año, mes y período de compra.
- Tiempo de entrega.
- Diferencia entre fecha real y fecha estimada de entrega.
- Indicador de entrega tardía.
- Cantidad de ítems.
- Cantidad de productos distintos.
- Cantidad de vendedores asociados.
- Valor total de productos.
- Valor total del flete.
- Valor total pagado.
- Cantidad de registros de pago.
- Número máximo de cuotas.
- Cantidad de medios de pago utilizados.
- Puntuación media de las reviews.
- Cantidad de reviews y comentarios.

Se verificó que la tabla mantuviera la granularidad esperada:

- 99.441 filas.
- 99.441 `order_id` únicos.

Esto confirma que los procesos de agregación y `merge` no generaron duplicación de pedidos.

## `order_items_analysis`

Granularidad:

`1 fila = 1 ítem vendido`

Esta tabla parte de `order_items` y se complementa con información de productos, categorías, vendedores, pedidos y clientes.

Incluye:

- `order_id` y `order_item_id`.
- Producto.
- Categoría del producto.
- Precio.
- Flete.
- Valor total del ítem (`price + freight_value`).
- Peso y dimensiones del producto.
- Vendedor.
- Ciudad y estado del vendedor.
- Estado del pedido.
- Fecha y período de compra.
- Cliente y ubicación del cliente.

La clave compuesta `(order_id, order_item_id)` continúa siendo única luego de los `merge`.

Se verificó:

- 112.650 filas.
- 0 duplicados en `(order_id, order_item_id)`.

Esta tabla será utilizada principalmente para analizar productos, categorías, vendedores, precios, fletes y volumen de ventas.

## `customer_analysis`

Granularidad:

`1 fila = 1 cliente real`

La tabla se construyó agrupando `orders_analysis` por `customer_unique_id`.

Incluye:

- Cantidad total de pedidos.
- Valor total pagado.
- Ticket promedio.
- Primera compra.
- Última compra.
- Tiempo de vida observado del cliente.
- Tiempo promedio de entrega.
- Puntuación media de reviews.
- Indicador de cliente recurrente.

Se verificó:

- 96.096 clientes únicos.
- 96.096 filas.
- 2.997 clientes con más de una compra.

Esta tabla permitirá analizar recurrencia, gasto por cliente y comportamiento general de compra.

## Criterio de modelado

Las tres tablas mantienen granularidades diferentes y complementarias:

- `orders_analysis`: análisis a nivel pedido.
- `order_items_analysis`: análisis a nivel producto vendido.
- `customer_analysis`: análisis a nivel cliente.

Esta separación evita duplicaciones de métricas y permite utilizar la tabla adecuada según la pregunta de negocio que se quiera responder.
