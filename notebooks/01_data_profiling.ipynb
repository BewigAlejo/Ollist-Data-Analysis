```python
#asi se cargan las tablas en un diccionario, donde la clave es el nombre de la tabla y el valor es el DataFrame

tables = {}
for file in files:
    if file.endswith(".csv"):
        table_name = file.replace(".csv", "")
        table_path = os.path.join(path, file)
        tables[table_name] = pd.read_csv(table_path)
        print(f"Loaded table: {table_name}")

tables["olist_customers_dataset"].head()
```

```python
#ahora vamos a ver el tamanio de las tablas
for name, df in tables.items():
    print(f"Table: {name}, Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
#primer resumen basico automatico, seria un data quality overview, que nos da una idea de la calidad de los datos y de la estructura de las tablas
summary = {}

for name, df in tables.items():
    summary[name] = {
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_values": df.isnull().sum().sum(),
        "duplicate_rows": df.duplicated().sum(),
        "data_types": df.dtypes.value_counts().to_dict()
    }
summary_df = pd.DataFrame.from_dict(summary, orient='index')
summary_df
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')

orders[date_columns].dtypes
```

Conversión de columnas temporales a tipo datetime
Las columnas relacionadas con fechas y horas se encontraban originalmente almacenadas como tipo str. Para poder realizar operaciones temporales, comparaciones cronológicas y cálculos de duración, se convirtieron al tipo datetime utilizando pd.to_datetime().

Se transformaron las siguientes columnas: order_purchase_timestamp, order_approved_at, order_delivered_carrier_date, order_delivered_customer_date y order_estimated_delivery_date.

Se utilizó el parámetro errors="coerce" para que cualquier valor no interpretable como fecha sea convertido en NaT, preservando así los valores faltantes dentro de un tipo de dato temporal adecuado. Luego de la conversión, se verificó que las cinco columnas quedaran almacenadas como datetime64[us].

Esta transformación permite continuar con controles de calidad más avanzados, como verificar el orden lógico de los eventos del pedido y calcular tiempos transcurridos entre las distintas etapas del proceso.

```python
orders["order_status"].value_counts()

# verificacion de la coherencia cronologica de las fechas, por ejemplo, que la fecha de aprobacion no sea posterior a la fecha de entrega al cliente
# purchase <= approved <= carrier <= customer
#regla 1 purchase <= approved


approved_before_purchase = orders[
    (orders["order_approved_at"] < orders["order_purchase_timestamp"])
]
approved_before_purchase.shape[0]
#el resultado es 0, lo que significa que no hay registros donde la fecha de aprobación sea anterior a la fecha de compra, lo cual es coherente.

#regla 2 approved <= carrier
carrier_before_approved = orders[
    (orders["order_delivered_carrier_date"] < orders["order_approved_at"])
]
carrier_before_approved.shape[0]
#el resultado es 1359, lo que significa que hay 1359 registros donde la fecha de entrega por el transportista sea anterior a la fecha de aprobación, lo cual es incoherente.

#regla 3 carrier <= customer
customer_before_carrier = orders[
    (orders["order_delivered_customer_date"] < orders["order_delivered_carrier_date"])
]
customer_before_carrier.shape[0]
#el resultado es 23, lo que significa que hay 23 registros donde la fecha de entrega al cliente sea anterior a la fecha de entrega por el transportista, lo cual es incoherente.


carrier_before_approved = orders[
    orders["order_delivered_carrier_date"] < orders["order_approved_at"]
].copy()

carrier_before_approved["time_difference"] = (
    carrier_before_approved["order_approved_at"]
    - carrier_before_approved["order_delivered_carrier_date"]
)
carrier_before_approved["time_difference"].describe()
```


Se verificó la coherencia cronológica entre las principales fechas del proceso de los pedidos, considerando como secuencia esperada purchase <= approved <= carrier <= customer. No se encontraron pedidos cuya fecha de aprobación fuera anterior a la fecha de compra. Sin embargo, se detectaron 1359 registros donde la fecha de entrega al transportista es anterior a la aprobación del pedido y 23 registros donde la fecha de entrega al cliente es anterior a la fecha registrada de entrega al transportista. Estos casos se consideran anomalías cronológicas potenciales y requieren un análisis adicional de las diferencias temporales antes de clasificarlos definitivamente como errores de calidad de datos.

Se identificaron 1359 pedidos cuya fecha de entrega al transportista es anterior a la fecha de aprobación. Las diferencias presentan una alta dispersión: van desde 21 segundos hasta aproximadamente 171 días. La mediana es de aproximadamente 17 horas y el 75 % de los casos presenta una diferencia inferior a aproximadamente 1 día y 2 horas. La existencia de valores extremos considerablemente mayores sugiere que las anomalías podrían tener distintas causas, por lo que es necesario analizar la distribución de las diferencias antes de determinar su tratamiento.
